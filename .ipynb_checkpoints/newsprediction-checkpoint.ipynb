```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('news.csv' , index_col=0)
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['content'] = df['title'] + "  " + df['text']
df = df.drop(columns = ['title' , 'text'])
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
```

```python
df['label'] = encoder.fit_transform(df['label'])
```

```python
df
```

```python
df['content'] = df['content'].apply(lambda x : x.lower())
```

```python
import string
```

```python
def remove_punc(txt):
    return txt.translate(str.maketrans('' , '' , string.punctuation))
```

```python
df['content'] = df['content'].apply(remove_punc)
```

```python
def remove_num(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new
```

```python
df['content'] = df['content'].apply(remove_num)
```

```python
df
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
def remove_stop(txt):
    words = word_tokenize(txt)
    cleaned_txt = []
    for i in words:
        if not i in stop_words:
            cleaned_txt.append(i)
    return ' '.join(cleaned_txt)
```

```python
df['content'] = df['content'].apply(remove_stop)
```

```python
df
```

```python
X = df['content']
```

```python
y = df['label']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.20 , random_state = 42)
```

## Using Bag of words Vectorizer and Naive Bias

```python
from sklearn.feature_extraction.text import TfidfVectorizer , CountVectorizer
```

```python
bow_vectorizer = CountVectorizer()
```

```python
X_train_bow = bow_vectorizer.fit_transform(X_train)
```

```python
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
model_nb = MultinomialNB()
```

```python
model_nb.fit(X_train_bow , y_train)
```

```python
y_pred_nb = model_nb.predict(X_test_bow)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_pred_nb , y_test)
```

## Using Bag of Words Vectorizer and Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
model_LR = LogisticRegression()
```

```python
model_LR.fit(X_train_bow , y_train)
```

```python
y_pred_LR = model_LR.predict(X_test_bow)
```

```python
accuracy_score(y_pred_LR , y_test)
```

## Using TFID Vectorizer and Naive Bias

```python
tf_vectorizer = TfidfVectorizer()
```

```python
X_train_tf = tf_vectorizer.fit_transform(X_train)
X_test_tf = tf_vectorizer.transform(X_test)
```

```python
model_nb2 = MultinomialNB()
```

```python
model_nb2.fit(X_train_tf , y_train)
```

```python
y_pred_nb2 = model_nb2.predict(X_test_tf)
```

```python
accuracy_score(y_pred_nb2 , y_test)
```

## Using TFID Vectorizer and Logistic Regression

```python
model_LR2 = LogisticRegression()
```

```python
model_LR2.fit(X_train_tf , y_train)
```

```python
y_pred_LR2 = model_LR2.predict(X_test_tf)
```

```python
accuracy_score(y_pred_LR2 , y_test)
```
